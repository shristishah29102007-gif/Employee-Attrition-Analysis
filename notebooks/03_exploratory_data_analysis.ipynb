```python
pd.crosstab(df['OverTime'], df['Attrition'])
```

```python
pd.crosstab(
    df['OverTime'],
    df['Attrition'],
    normalize='index'
) * 100
```

```python
from scipy.stats import chi2_contingency

contingency_table = pd.crosstab(df['OverTime'], df['Attrition'])

chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("Chi-Square Statistic:", chi2)
print("P-value:", p_value)
print("Degrees of Freedom:", dof)
```

### Chi-Square Test: OverTime vs Attrition

**Hypotheses:**

- **Null Hypothesis (H₀):** There is no significant association between overtime and employee attrition.
- **Alternative Hypothesis (H₁):** There is a significant association between overtime and employee attrition.

**Test Results:**

- Chi-Square Statistic: 87.5643
- P-value: 8.1584 × 10⁻²¹
- Degrees of Freedom: 1
- Significance Level: 0.05

**Interpretation:**

Since the p-value is less than 0.05, we reject the null hypothesis. Therefore, there is a statistically significant association between overtime and employee attrition.

Employees who work overtime have a higher observed attrition rate (30.53%) compared with employees who do not work overtime (10.44%).

**Conclusion:** Overtime is significantly associated with employee attrition in this dataset. However, this statistical result does not establish that overtime causes employees to leave.

### Attrition Rate by Job Satisfaction - Key Observations

The bar chart displays the percentage of employees who left the organization, grouped by their Job Satisfaction Level (1 = Low to 4 = Very High).

1. **General Decreasing Trend:** The attrition rate generally decreases as job satisfaction increases, although there is a very small increase between Levels 2 and 3.
2. **Highest Attrition at Level 1 (Low):** Employees with low job satisfaction have the highest attrition rate of 22.84%.
3. **Similar Attrition at Levels 2 and 3:** Employees with Medium and High job satisfaction have nearly identical attrition rates of 16.43% and 16.52%, respectively.
4. **Lowest Attrition at Level 4 (Very High):** Employees with very high job satisfaction have the lowest attrition rate of 11.33%.

**Key takeaway:** Employees with low job satisfaction have an attrition rate approximately twice as high as employees with very high job satisfaction. This suggests an association between job satisfaction and employee attrition, which we will examine further using the Chi-Square test.

```python
from scipy.stats import chi2_contingency

contingency_table = pd.crosstab(
    df['JobSatisfaction'],
    df['Attrition']
)

chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("Chi-Square Statistic:", chi2)
print("P-value:", p_value)
print("Degrees of Freedom:", dof)
```

### Chi-Square Test: JobSatisfaction vs Attrition

**Hypotheses:**

- **Null Hypothesis (H₀):** There is no significant association between job satisfaction and employee attrition.
- **Alternative Hypothesis (H₁):** There is a significant association between job satisfaction and employee attrition.

**Test Results:**

- Chi-Square Statistic: 17.5051
- P-value: 0.0005563
- Degrees of Freedom: 3
- Significance Level: 0.05

**Interpretation:**

Since the p-value (0.0005563) is less than the significance level (0.05), we reject the null hypothesis.

Therefore, there is a statistically significant association between job satisfaction and employee attrition.

The observed attrition rate is highest among employees with low job satisfaction (22.84%) and lowest among employees with very high job satisfaction (11.33%).

**Conclusion:** Job satisfaction is significantly associated with employee attrition in this dataset. However, this result does not establish a cause-and-effect relationship.

### Years at Company vs Attrition - Key Observations

The box plot compares employee tenure (YearsAtCompany) between employees who left the organization and those who stayed.

1. **Lower Median Tenure:** Employees who left the organization have a lower median tenure (approximately 3 years) compared with employees who stayed (approximately 6 years).
2. **Distribution Spread:**
   - For employees who left, the middle 50% of tenure values are approximately between 1 and 7 years.
   - For employees who stayed, the middle 50% are approximately between 3 and 10 years.
3. **Presence of Outliers:** Both groups contain outliers, indicating that some employees have considerably longer tenure than others in their respective groups.
4. **Average Tenure Comparison:** The mean tenure of employees who left (5.13 years) is lower than that of employees who stayed (7.37 years).

**Key takeaway:** Employees who left generally had shorter tenure than employees who stayed. We will use an independent samples t-test to determine whether the difference in average tenure is statistically significant.

```python
from scipy.stats import ttest_ind

stayed = df[df['Attrition'] == 'No']['YearsAtCompany']
left = df[df['Attrition'] == 'Yes']['YearsAtCompany']

t_stat, p_value = ttest_ind(
    stayed,
    left,
    equal_var=False
)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

### Independent Samples t-test: YearsAtCompany vs Attrition

**Hypotheses:**

- **Null Hypothesis (H₀):** There is no significant difference in average tenure between employees who stayed and employees who left.
- **Alternative Hypothesis (H₁):** There is a significant difference in average tenure between the two groups.

**Test Results:**

- T-statistic: 5.2826
- P-value: 2.2859 × 10⁻⁷
- Significance Level: 0.05
- Test Used: Welch's Independent Samples t-test

**Interpretation:**

Since the p-value is less than 0.05, we reject the null hypothesis. Therefore, there is a statistically significant difference in average tenure between employees who stayed and employees who left.

Employees who stayed had an average tenure of approximately 7.37 years, whereas employees who left had an average tenure of approximately 5.13 years.

**Conclusion:** Employees who left the organization had significantly shorter average tenure than employees who stayed. This indicates an association between tenure and employee attrition, but it does not establish causation.
